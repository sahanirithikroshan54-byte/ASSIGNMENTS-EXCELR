```python
#1.Program for Prime numbers
num=int(input("Enter a number:"))
if num>1:
    for i in range(2,num):
        if num%i==0:
            print(num,"Number is not a prime number")
            break
    else:
            print(num,"is a prime number ")
else:
    print("Number is less than 1")
```

```python
## 2.Product of Random Numbers
import random
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)
print("num1 is :",num1)
print("num2 is :",num2)
user_answer = int(input(f"Product of random numbers {num1} * {num2}:"))
correct_answer = num1 * num2
if user_answer == correct_answer:
    print("Correct answer!")
else:
    print("Wrong answer.")
    print("The correct answer is:", correct_answer)
```

```python
#4.Hospital Billing System
stay= int(int(input("Number of days stayed :")))
Room_charge_per_day = 2000
stay_in_room=(Room_charge_per_day * stay)
Doctor_consultation_fee = 1500
Test=int(input("Number of tests :"))
Lab_test_charge = 300
Lab_charge=(Lab_test_charge * Test)
Medicine_charge=int(input("Enter the Medicine charges :"))
Total_charges=stay_in_room + Doctor_consultation_fee +Lab_charge + Medicine_charge
print(f"Total charge is : {Total_charges}")
if Total_charges >= 10000:
    Discount=Total_charges*0.10
    Discount_bill = Total_charges-Discount
    print(f"After Discount total bill is: {Discount_bill}")
else:
    print(f"Your final amount is : {Total_charges}")
```
