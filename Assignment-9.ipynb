1.Standard scaler:Converts features to mean 0 and standard deviation 1. Commonly used with Logistic Regression, SVM, KNN, PCA, and Neural Networks.
Preferred Scenario:
Data is approximately normally distributed or has varying ranges.
2.Minmaxscaler:Preserves the relative distribution of data and is useful for Neural Networks and distance-based algorithms.
Preffered scenario:
Data has a known bounded range and you want values between 0 and 1.

One-Hot Encoding:
pros:
• Does not create an artificial order between categories.
• Works well for nominal data such as colors or cities.
• Easy for models to interpret.
cons:
• Can create many columns when there are many categories.
• Increases memory usage and dataset size.
• May cause the curse of dimentionality
Label Encoding:
pros:
• Simple and easy to implement.
• Creates only one additional numerical column.
• Uses less memory than one-hot encoding.
• Useful for ordinal categories.
cons:
• Can introduce a false sense of order between categories.
• Numerical values may be incorrectly interpreted as having meaningful differences.
• Generally unsuitable for nominal categories with no natural order.

```python
# num_cols = df.select_dtypes(include=['int64', 'float64']).columns
num_cols = ['age','fnlwgt','education_num','capital_gain',
                  'capital_loss','hours_per_week']
for col in num_cols:
    print(col,':', df[col].skew())
```

```python
# features 1
df['total_capital'] = df['capital_gain'] - df['capital_loss']
df['total_capital']
```

```python
# features 2
df['total_time'] = df['hours_per_week'] / df['age']
df['total_time']
```

```python
# skewness
real_num_cols = ['age','fnlwgt','education_num','capital_gain',
                 'capital_loss','hours_per_week',
                 'total_capital','total_time']

for col in real_num_cols:
    print(col, df[col].skew())
```

```python
# log function
df['capital_gain_log'] = np.log1p(df['capital_gain'])
df['capital_loss_log'] = np.log1p(df['capital_loss'])
df['total_capital_log'] = np.log1p(df['total_capital'])
df['fnlwgt_log'] = np.log1p(df['fnlwgt'])
```

```python
# skewness

num_cols = ['age','education_num','hours_per_week','total_time',
                 'capital_gain_log','capital_loss_log','total_capital_log','fnlwgt_log']

for col in num_cols:
    print(col, df[col].skew())
```

Skewness analysis was performed on  features , and the log functions are applied on high skewness features,but the features like capital_loss_log changes moderately.

Thus the transformation improved the data for modelling pupose.
