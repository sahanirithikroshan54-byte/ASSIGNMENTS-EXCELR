```python
from sklearn.impute import SimpleImputer
num_cols = df.select_dtypes(include=['float']).columns
df[num_cols] = SimpleImputer(strategy='median').fit_transform(df[num_cols])
```

```python
df=df.drop_duplicates()
df
```

```python
for col in df.select_dtypes(include=['float']).columns:
    sns.boxplot(df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
for i in df.select_dtypes(include=['float']).columns:
    q1=df[i].quantile(0.25)
    q3=df[i].quantile(0.75)
    iqr=q3-q1
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr
    df[i]=df[i].clip(lower,upper)
    print("inter quartile range is:",iqr)
    print("lower bound is:",lower)
    print("upper bound is:",upper)
```

```python
for col in df.select_dtypes(include=['float']).columns:
    sns.boxplot(df[col])
    plt.title(f"boxplot of {col}")
    plt.show()
```

4. Pattern Recognition and Insights:
A.Correlation between Variables:
ASTV (Abnormal Short-Term Variability) shows the strongest positive correlation with the fetal state (NSP), making it an important indicator of abnormal fetal health.
ALTV (Abnormal Long-Term Variability) also has a moderate positive correlation with NSP, indicating its significance in predicting fetal condition.
AC (Accelerations) has a moderate negative correlation with NSP, suggesting that higher accelerations are generally associated with healthier fetal conditions.
DP (Prolonged Decelerations) shows a weak-to-moderate positive correlation, indicating increased risk when decelerations are higher.
Other variables such as LB, UC, MLTV, FM, Width, and Tendency have weak correlations, meaning they have a smaller individual impact on fetal health.
B.Trends Over Time:
The dataset does not contain any time or date attribute.
Therefore, temporal trends or time-based pattern analysis cannot be performed.
The analysis is limited to identifying relationships and patterns among the available variables.

5.Overall Conclusion:
A.Key Patterns Identified:
ASTV and ALTV are the most influential variables for identifying fetal health status.
Higher values of ASTV and ALTV are generally associated with abnormal fetal conditions.
Higher acceleration values (AC) indicate better fetal well-being.
No variables show extremely high correlations with each other, indicating low multicollinearity and that each feature contributes unique information.
B.Decision Making impact:
ASTV and ALTV are the strongest predictors of fetal health condition.
Accelerations (AC) are positively associated with normal fetal health and negatively associated with abnormal conditions.
Most variables contribute only weakly on their own, indicating that combining multiple features is more effective for predicting fetal health.
Since no temporal data is available, the analysis focuses solely on correlations and feature relationships rather than trends over time.
