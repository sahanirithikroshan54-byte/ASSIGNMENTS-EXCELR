1.State the hypothesis statements:
Equation W=100 + 5 * x |W=100 +5 * 600=4000(x=600)
Null Hypothesis=Weekly cost estimated through the usual model prediction is sufficient.
(H0) = 4000
Alternative Hypothesis=Weekly cost estimated is higher than usual model prediction.
(Ha) = Ha > 4000

```python
#2. Calculate the Test Statistic:(z-Test)
from statsmodels.stats.weightstats import ztest
from scipy.stats import norm
x_bar=3050
pop_mean=4000
pop_std=5*25
n=25
alpha=0.05
z_calculated=(x_bar-pop_mean)/(pop_std/np.sqrt(n))
p_value = 1 - stats.norm.cdf(z_calculated)
z_critical = norm.ppf(1 - alpha)
print("z_calculated value is:",z_calculated)
print(f"P_value: {p_value:.3f}")
print(f"z_critical value is:{z_critical}")
```

```python
#3. Determine the Probability and compare:
if p_value < alpha:
    print("Reject The Null Hypothesis.")
else:
    print("Fail To Reject Null Hypothesis.")
```

```python
#4. Make a Decision:
alpha=0.05
if z_calculated < z_critical:
    print("Fail To Reject Null Hypothesis.")
else:
    print("Reject Null Hypothesis.")
```

```python
#Confidence intervals
#A.Build 99% Confidence Interval Using Sample Standard Deviation
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

# Sample size
n = len(data)

# Sample mean
mean = np.mean(data)

# Sample standard deviation
std = np.std(data, ddof=1)

# Standard error
se = std / np.sqrt(n)

# Confidence level
confidence = 0.99

# t critical value
t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)

# Margin of error
margin_error = t_critical * se

# Confidence interval
lower = mean - margin_error
upper = mean + margin_error

print("Sample Mean:",mean)
print("Sample Standard Deviation:",std)
print("t-Critical Value:",t_critical)
print("Margin of Error:",margin_error)
print("99% Confidence Interval:",(lower, upper))
```

```python
#B.Build 99% Confidence Interval Using Known Population Standard Deviation
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 
        1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

# Sample mean
mean = np.mean(data)

# Known population standard deviation
sigma = 0.2

# Sample size
n = len(data)

# Z critical value for 99% confidence
z = norm.ppf(0.995)

# Standard Error
se = sigma / np.sqrt(n)

# Margin of Error
me = z * se

# Confidence Interval
lower = mean - me
upper = mean + me

print("Sample Mean:",mean)
print("99% Confidence Interval:",(lower, upper))
```
