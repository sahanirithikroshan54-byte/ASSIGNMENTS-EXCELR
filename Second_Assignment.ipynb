```python
#Section A:
#1.Various operations on list
List=[10,20,10,30,40]
print(List)
List.append(50)
print(List)
List.extend([60])
print(List)
List.insert(3,20)
print(List)
List.remove(10)
print(List)
List.pop()
print(List)
List.pop(2)
print(List)
print(List.index(30))
print(List.count(10))
List.sort()
print(List)
List.reverse()
print(List)
List.clear()
print(List)
```

```python
#2.Tuple 
Tuple=("Rithik","Roshan","Sahani")
print(Tuple)
Tuple[1]="Kohli"
#Ans:Tuple is a data structure, where it  does not support/allow to change the elements after they are assigned.
#These tuples are immutable and cannot be modified or altered after assigning elements.
```

```python
#3.Set
Set={1,2,3,4,2,1,5}
print(Set)
#A set is a unordered collection of data. where it does not aloow any duplicate values in set, which means values should be unique. 
#The set is mutable but allows no duplication of elements and unordered. 
```

```python
#5.Eligibile for vote
age=int(input("Enter the age:"))
if (age >= 18):
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
#6.Grading mechanism
Marks=int(input("Enter the marks :"))
if Marks >= 90:
    print("A Grade")
elif Marks >= 75:
    print("B Grade")
elif Marks >= 50:
    print("C Grade")
else:
    print("Fail")
```

```python
#7.Program on Numbers
num=int(input("Enter a number:"))
if num > 1:
    if num % 2 == 0:
        print("It is a positive and Even Number")
    else:
        print("Not an even number")
elif num == 0:
    print("It is zero")
else:
    print("It is a Negative number")
```
